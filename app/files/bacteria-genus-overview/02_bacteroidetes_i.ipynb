# Bacteroidetes I: *Bacteroides* & *Prevotella*

The single most reproducible finding in gut microbiome research, and the
two genera behind it. Both are Bacteroidetes — both break down complex
carbohydrates the human body can't digest on its own — but they
specialize in different ones, which is exactly why diet splits them so
cleanly.

## *Bacteroides*

Thrives on the fats and proteins common in a "Western" diet, and on
processed carbohydrates. Extremely abundant in industrialized-country
guts. Mostly beneficial and commensal, but the genus has a clinically
important exception: *Bacteroides fragilis*, usually harmless, is also
one of the most common anaerobic bacteria isolated from intra-abdominal
infections when gut contents escape into the body cavity (e.g. after a
ruptured appendix) — a reminder that "genus" can span both a beneficial
majority and a clinically relevant minority (the theme of notebook 11).

## *Prevotella*

Specializes in breaking down plant fiber — cellulose, xylan, the
carbohydrates common in high-fiber, plant-heavy diets. Dominant in
agrarian and non-industrialized populations. One specific species,
*Prevotella copri*, has been studied for a possible association with
rheumatoid arthritis in some cohorts — an active research area, and, like
everything correlational in this course, not evidence that *Prevotella*
causes the condition.

## Enterotypes: a real (if contested) clustering pattern

Some researchers have proposed that human gut microbiomes cluster into a
small number of "enterotypes," with *Bacteroides*-dominant vs.
*Prevotella*-dominant being the most cited split. The idea is debated —
critics argue it's more of a continuous gradient than discrete clusters —
but the underlying data pattern (few people are high in both at once) is
real, and easy to see for yourself.

## 1. Load the real data

```python
import pandas as pd

df = pd.read_csv("../gut-microbiome-under-glass/real_gut_usa_malawi.csv")
df[["country", "Bacteroides", "Prevotella"]].groupby("country").mean().round(1)
```

## 2. A Bacteroides/Prevotella ratio, per sample

```python
df["bact_prev_ratio"] = df["Bacteroides"] / (df["Prevotella"] + 0.01)  # +0.01 avoids divide-by-zero
df.groupby("country")["bact_prev_ratio"].median().round(1)
```
