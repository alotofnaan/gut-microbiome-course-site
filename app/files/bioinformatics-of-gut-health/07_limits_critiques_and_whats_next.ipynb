# Limits, Critiques, and What's Next

Every notebook so far has built up how GMWI2 works. This last one is
about where it stops working — the same skeptical habit the other
course's p-value notebooks trained: understanding a method well enough to
know exactly what it can't tell you.

## 1. Disease-agnostic, not diagnostic

GMWI2 was trained across dozens of *unrelated* diseases pooled together —
that's a deliberate design choice, not an oversight. It means a low score
signals "this profile resembles the disease-associated samples in
training," not "you have condition X." Treat it the way the other
course's notebooks 03-04 taught you to treat any single p-value: informative,
not diagnostic, and much more useful for tracking *change* in one person
over time than for asserting anything from one score in isolation.

## 2. Generalization depends on who's in the training set

Recall the other course's real finding (notebook 05, USA vs. Malawi):
*Prevotella* and *Bacteroides* trade dominance almost entirely based on
diet and geography, not disease. A model trained mostly on samples from
one kind of population risks learning "this looks like the training
country" rather than "this looks unhealthy" — which is exactly why
GMWI2's authors deliberately pooled 54 studies across 26 countries and six
continents rather than training on a single-country dataset. This is a
real, active concern in the field, not a hypothetical: microbiome-based
classifiers have a documented history of generalizing worse across
populations than papers' headline accuracy numbers suggest, precisely
because of effects like the diet/geography one you already reproduced
yourself in the other course.

## 3. Presence/absence throws away real information

Notebook 03 asked you to think about why collapsing abundance into a
binary presence/absence signal might be deliberate. One real answer:
relative abundance is noisy and compositional (the other course's
notebook 01, section 7) — presence/absence is a cruder but more robust
signal to a model, at the cost of ignoring genuine differences between
"barely detectable" and "dominant."

## 4. Where the field is headed

- **PICRUSt2**-style functional prediction (mentioned in the other
  course's notebook 07) is one direction: not just "who's there" or "is
  the community healthy-shaped," but "what genes/pathways are likely
  present" — with the same predicted-not-measured caveat.
- **Larger, more diverse pooled datasets** are the direct answer to
  concern #2 above — expect health-index papers to keep growing their
  training cohorts geographically, the same way GMWI2 improved on the
  original GMHI.
- **Longitudinal tracking** (the same person's GMWI2 score over time,
  rather than one score in isolation) plays to the method's actual
  strength — noticing *change* — rather than its weakness (one-shot
  diagnosis).
