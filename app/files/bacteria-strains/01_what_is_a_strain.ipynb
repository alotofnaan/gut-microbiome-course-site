# What Is a Strain?

Notebook 01 of the species course anchored species to a number: ANI
around 95-96%. Strain doesn't get an equally clean number — and that's
the point of starting here.

## The working definition

A **strain** is a genetically distinct lineage descended from a common
ancestor within one species — distinguished by specific mutations, gene
gains/losses, or acquired elements (notebook 05), not by crossing some
genome-wide similarity threshold the way species roughly does. Two
*E. coli* strains can be well above 99% identical genome-wide and still
be meaningfully, sometimes dangerously, different — because the
difference that matters is often a handful of specific genes, not overall
similarity.

## Why this resolution level exists at all

Species-level data (the whole previous course) answers "which organism is
this." Strain-level data answers a different question: "which *version*
of this organism, and does that version carry the specific
trait I actually care about?" The species course's notebook 06 made this
concrete: *E. coli* the species contains both harmless commensals and
Shiga-toxin-producing pathogens. Species-level sequencing, even done
perfectly, cannot tell those apart. Strain-level analysis exists
specifically to answer that question.
