# Firmicutes II: *Ruminococcus*, *Blautia*, *Dorea*, *Lachnospira*, *Oscillospira*

Five more Firmicutes, all broadly in the fiber-fermenting, generally
beneficial camp — the "supporting cast" of a healthy gut that rarely gets
individually named in casual writing about the microbiome, but that make
up a large fraction of it by abundance.

## The lineup

- ***Ruminococcus*** — fiber/cellulose degrader, named for its similarity
  to bacteria found in ruminant (cattle, sheep) guts.
- **`[Ruminococcus]`** — the bracketed, provisionally-classified cousin
  from notebook 01. Tracked separately in this dataset for a reason.
- ***Blautia*** — one of the more recently defined genera (named 2008);
  broadly fermentative, frequently detected across many human populations.
- ***Dorea*** — another Lachnospiraceae-family fermenter, often
  co-occurring with *Blautia* and *Roseburia*.
- ***Lachnospira*** — pectin-degrading specialist (pectin is a fiber
  found in fruit).
- ***Oscillospira*** — historically hard to culture in a lab (culturing
  didn't succeed until 2016, despite the genus being described over a
  century earlier from microscopy alone) — a good reminder that
  "detected by sequencing" and "grown in a lab" are different milestones.

## 1. Load and compare all five across countries

```python
import pandas as pd

df = pd.read_csv("../gut-microbiome-under-glass/real_gut_usa_malawi.csv")
cols = ["Ruminococcus", "[Ruminococcus]", "Blautia", "Dorea", "Lachnospira", "Oscillospira"]
df.groupby("country")[cols].mean().round(2).T
```

## 2. Visualize all five at once

```python
import matplotlib.pyplot as plt

means = df.groupby("country")[cols].mean()
fig, ax = plt.subplots(figsize=(7, 4))
means.T.plot(kind="bar", ax=ax, color=["#4C72B0", "#DD8452"])
ax.set_ylabel("Mean relative abundance (%)")
ax.set_title("Five Firmicutes genera, USA vs. Malawi")
plt.xticks(rotation=30, ha="right")
fig.tight_layout(); plt.show()
```
