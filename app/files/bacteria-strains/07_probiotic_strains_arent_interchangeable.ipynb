# Probiotic Strains Aren't Interchangeable

The species course's notebook 04 covered *Bifidobacterium* species —
*adolescentis* vs. the infant-associated *longum*/*infantis*. Even that
was one level too coarse for how probiotic research actually works.

## Clinical evidence is strain-specific, not species-specific

Two real, well-studied examples:

- ***Lactobacillus rhamnosus* GG (LGG)** — one of the most-studied
  probiotic strains in the world, with a genuine body of clinical trial
  evidence for specific uses (e.g., certain kinds of diarrhea).
- ***Bifidobacterium longum* BB536** — a specific, commercially used
  *B. longum* strain with its own separate clinical trial record.

The critical point: a clinical trial result for LGG specifically **does
not automatically transfer** to a different *Lactobacillus rhamnosus*
strain, even though they're the same species — because the strain-specific
genes each one happens to carry (adhesion factors, metabolic
byproducts, immune-signaling molecules) are exactly the kind of trait
this whole course has been about. A supplement label that just says the
species name, with no strain designation, is telling you less than it
sounds like.

## Why this is a strain question, not a marketing detail

This directly extends notebook 01's definition: a probiotic's clinical
effect (if any) plausibly depends on specific genes — surface proteins,
metabolic pathways, sometimes plasmid-borne elements (notebook 05) — that
are strain properties, not automatic species-wide guarantees. This is
also why serious probiotic research papers name strains precisely
(genus, species, AND a strain designation like "GG" or "BB536") rather
than stopping at species.
