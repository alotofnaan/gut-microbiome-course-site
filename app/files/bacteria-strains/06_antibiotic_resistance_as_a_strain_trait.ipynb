The genus course's notebook 07 flagged carbapenem-resistant *Klebsiella

one level deeper: resistance is fundamentally a **strain-level**, often
**plasmid-borne** (notebook 05) trait, not a species-level one.

## `blaKPC`: a real, specific gene

**KPC (*Klebsiella pneumoniae* carbapenemase)** is a real enzyme —
encoded by the `blaKPC` gene — that breaks down carbapenem antibiotics,
often considered a "last-resort" drug class. A *Klebsiella pneumoniae*
strain (or, via plasmid transfer, a strain of a completely different
species) carrying `blaKPC` can survive treatment that would kill a
non-resistant strain of the exact same species. This is precisely why
"the bacterium is *Klebsiella pneumoniae*" (species-level) and "the
bacterium is treatable with carbapenems" (strain-level, resistance-gene
dependent) are two different, separately-important facts.
