# Welcome — How to Use This Course

This course runs entirely **in your browser**. There's nothing to install —
the notebooks you're reading are real, live Python, executing on your machine
via a technology called Pyodide (Python compiled to WebAssembly). No server,
no video, no simulation. When you run a cell, real code runs.

The first time a notebook loads, give it a few seconds — it's downloading a
real Python interpreter into your browser tab.

## The one skill you need: running a cell

Click into any gray code cell and press **Shift+Enter** (or the ▶ button).
That's it. The cell runs, and any output (numbers, tables, plots) appears
right below it.

Cells run **top to bottom** — later cells often depend on variables created
by earlier ones. If something errors, try re-running the notebook from the
top (menu: Run → Run All Cells).

## The pattern every notebook follows

Every notebook here uses the same loop:

1. **Read** the markdown explanation.
2. **Run** the code cell below it (Shift+Enter).
3. **Modify** it yourself in a "🔧 YOUR TURN" cell.
4. **Explain** what you saw in your own words in an "EXPLAIN" cell.

Steps 3 and 4 are where the actual learning happens — don't skip them, even
though nothing forces you to fill them in.
