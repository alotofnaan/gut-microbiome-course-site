# What Makes a Species?

Animals get a relatively clean species definition (can they interbreed
and produce fertile offspring?). Bacteria reproduce asexually and swap
genes across what look like species lines via horizontal gene transfer —
so the usual definition doesn't apply. This notebook covers what actually
gets used instead.

## The modern answer: Average Nucleotide Identity (ANI)

Compare two bacterial genomes base-by-base across their shared genes and
compute the percentage that match — that's **ANI**. The field has
converged on a rough convention: genomes with **≥ 95-96% ANI** are
usually called the same species; below that, different species. This
threshold isn't a law of nature — it was chosen because it closely
matched an older, slower method (DNA-DNA hybridization) that defined most
of today's named species before genome sequencing existed. It's a
convention chosen for continuity with history, not a discovered natural
boundary.

## Why 16S sequencing (courses 1 and 3) can't get you here

16S rRNA sequencing reads roughly 250-500 bases of one gene. Two
genuinely different species can share 97%+ identity in that one gene
while differing enormously everywhere else in their genomes — 16S simply
doesn't carry enough information to apply an ANI-style test. This is why
course 1's and course 3's data stops at genus, and why the bioinformatics
course's `toy_species_abundance.csv` — built on the idea of shotgun
metagenomics (course 2, notebook 02) — can carry species-level names at
all.

## A preview of a bigger problem

Because ANI is a **convention** (a chosen cutoff), not a discovered
boundary, two genomes at 94% ANI and two at 96% ANI aren't necessarily
biologically different in kind — they're on opposite sides of a line
