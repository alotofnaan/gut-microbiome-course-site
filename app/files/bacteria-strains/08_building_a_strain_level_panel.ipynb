# Building a Strain-Level Presence/Absence Panel

Same method as the bioinformatics course's notebook 03 (binarizing a
species table), one resolution level down: here, the "features" are
genes, and each row is a strain, not a sample of mixed species.

## 1. The panel is already binary

Unlike the bioinformatics course's abundance data, this course's genes
are recorded directly as presence/absence (0/1) — there's no abundance
value to threshold, since a gene is either in a genome or it isn't (no
partial presence at this resolution).

```python
import pandas as pd

df = pd.read_csv("ecoli_strain_profiles.csv")
genes = ["stx1", "stx2", "eae", "papC", "hlyA", "blaKPC"]
df[genes].sum(axis=0)
```

## 2. How many genes does a typical strain carry?

```python
df["n_genes_present"] = df[genes].sum(axis=1)
df.groupby("pathotype")["n_genes_present"].describe()[["mean", "std", "min", "max"]].round(2)
```
