# Species Isn't the End Either

Twelve modules down from genus, and the pattern should look familiar by
now: every resolution level solves real problems and creates a new blind
spot exactly where the next level picks up.

## What this course actually established

- Species is measured by a **chosen convention** (ANI ~95-96%), not a
  discovered natural line — notebook 01, made concrete in notebook 09.
- Real reclassification is happening **right now** in named genera this
  site already covered — *Faecalibacterium* (notebook 03), *Prevotella*
  (notebook 07) — not hypothetically, but as ongoing genomic work.
- Species-level resolution has a real, sharp limit: *E. coli* the
  harmless commensal and *E. coli* O157:H7 are — by every method covered
  in this course — literally the same species (notebook 06).

## What's next: strains

That last point is the entire reason a **future course on this site will
cover bacterial strains specifically**. Strain-level analysis is what
distinguishes a harmless *E. coli* from a dangerous one, or one probiotic
*Bifidobacterium longum* product from another — using methods
(whole-genome comparison of specific genes, virulence factor detection)
that go beyond even the ANI/MLST toolkit from notebook 08.
