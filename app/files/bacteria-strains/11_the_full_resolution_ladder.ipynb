# The Full Resolution Ladder

Five courses, one consistent story, told at finer and finer resolution
each time. This notebook is the actual ending.

## The ladder, start to finish

| Level | Course | What it can tell you | What it can't |
|---|---|---|---|
| Community / phylum | *The Gut Microbiome, Under Glass* | Which broad groups dominate, diet/geography effects, real statistics for testing claims | Which specific organisms, precisely |
| Genus | *Bacteria Genus Overview* | ~20 genera's real biology, ecological roles, research associations | Species-level function within a genus |
| Species | *Species-Level Resolution* | ANI-defined organism identity, some functional specificity | Strain-level traits within one species |
| Strain | *Bacteria Strains* (this course) | Which specific genes a lineage carries — virulence, resistance | Whether those genes are actually active right now |
| Function / expression | *(beyond this site, for now)* | What's actually happening, right now, in a specific host | — |

Every level down required a harder, more expensive method (16S → shotgun
metagenomics → whole-genome ANI/MLST → cgMLST/SNP typing → RNA
sequencing) and traded breadth/cost for precision.

## The pattern that held across all five courses

Every single course on this site ended with the same honest move: state
plainly what its own resolution level *can't* tell you, and point at what
comes next. That's not a coincidence or a rhetorical device repeated for
effect — it's the actual, honest shape of how this field works. There is
no final resolution level after which everything is finally, completely
known. There's just the next question a finer method can answer, and a
new blind spot exactly where that method's limits begin.
