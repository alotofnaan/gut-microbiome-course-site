# Genus Isn't the Whole Story

Twelve modules, roughly twenty genera, real biology for each one. This
closing notebook is about the resolution genus-level data *doesn't* give
you — and what comes next.

## The gap this whole course has been working around

Notebook 07 already previewed it: *Escherichia coli* the harmless
commensal and *E. coli* O157:H7 the toxin-producing pathogen are the same
species — genus tells you even less. *Bacteroides fragilis* is mostly
beneficial, except for the toxigenic strains associated with intra-
abdominal infection. A genus-level abundance number, even a perfectly
accurate one, cannot distinguish these cases. The bioinformatics course's
notebook 02 made the same point about genus vs. species resolution in
sequencing methods; this course has made it about biology directly.

## Why courses (including this one) still teach at genus level

Not because it's the most informative level — because it's usually the
most *reliable* level a standard 16S survey can resolve to (course 1,
notebook 01), and because many real ecological patterns (like the
Bacteroides/Prevotella diet split) are genuinely visible at genus
resolution even though finer distinctions exist underneath. Genus-level
analysis is a real, useful tool with a real, specific blind spot — not a
mistake.

## What's next: strains

Below species is **strain** — genetically distinct lineages within one
species (like *E. coli* K-12, a harmless lab workhorse strain, versus
O157:H7). Strain-level differences can determine whether an organism is
harmless, beneficial, or dangerous, and detecting them reliably needs
different methods than the ones covered in this course or the
bioinformatics course (typically whole-genome assembly or strain-specific
marker genes, rather than the marker-gene/16S approaches covered so far).
**A future course on this site will cover bacterial strains
specifically** — this notebook is the deliberate hand-off point for it.
