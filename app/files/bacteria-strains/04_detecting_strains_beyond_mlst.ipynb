# Detecting Strains: Beyond MLST

The species course's notebook 08 introduced MLST — comparing a handful
of housekeeping genes. Strain-level work usually needs to go further.

## cgMLST: MLST, scaled way up

**Core genome MLST (cgMLST)** applies the same basic idea as classic
MLST (species course, notebook 08) but compares hundreds to thousands of
genes shared across a species' "core genome," instead of just 7. This is
real, current outbreak-surveillance infrastructure — public health
systems like PulseNet and platforms like EnteroBase use cgMLST-style
schemes to determine whether two illness cases share a genetically
near-identical strain, which is exactly how real foodborne outbreaks get
traced back to a single contaminated source.

## SNP-based typing

An alternative/complementary approach: align whole genomes and count
**single-nucleotide polymorphisms (SNPs)** — individual base differences
— between them. Two strains differing by only a handful of SNPs across
their entire genome are very likely the same outbreak strain; thousands
of SNPs apart, unlikely. (If this sounds familiar: the same logic used to
track how a virus lineage evolves and spreads, applied to bacteria.)

## Whole-genome phylogenetics

Comparing many strains' SNP profiles (or cgMLST profiles) against each
other lets researchers build a **phylogenetic tree** of strains within
one species — literally a family tree, showing which strains are close
relatives and which diverged long ago. This is how the *Faecalibacterium*
split (species course, notebook 03) gets investigated at finer
resolution, and how EHEC outbreak strains get traced to their likely
common source.
