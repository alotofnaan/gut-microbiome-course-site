# *Akkermansia* Beyond *muciniphila*

The genus course's notebook 08 treated *Akkermansia* as essentially a
one-species genus. That was a reasonable simplification for a genus

## More than one member

*Akkermansia muciniphila* (described 2004) is still the by-far
best-studied species and the one nearly every metabolic-health study
refers to. But the genus has since gained additional described species —
for example *Akkermansia glycaniphila*, first described from the gut of a
reticulated python, and ongoing genomic surveys of human gut samples have
proposed further distinct *Akkermansia* clades/species beyond the classic
one. The practical takeaway: a 16S or genus-level "Akkermansia" reading
(as in the genus course) can't tell you which species — or whether more
than one — is actually present.
