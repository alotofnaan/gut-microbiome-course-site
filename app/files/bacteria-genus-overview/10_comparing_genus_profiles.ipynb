# Comparing Genus Profiles Across All Three Datasets

Every notebook so far pulled from one or two datasets at a time. This one
loads all three at once and asks a question none of the individual
courses asked: **which genera show up where, and how do the numbers
actually compare once you line them up?**

## 1. Load everything

```python
import pandas as pd

synthetic = pd.read_csv("../gut-microbiome-under-glass/example_gut_samples.csv")
real = pd.read_csv("../gut-microbiome-under-glass/real_gut_usa_malawi.csv")
toy = pd.read_csv("../bioinformatics-of-gut-health/toy_species_abundance.csv")

for name, d in [("example_gut_samples", synthetic), ("real_gut_usa_malawi", real), ("toy_species_abundance", toy)]:
    print(f"{name}: {d.shape[0]} samples, columns = {list(d.columns)}")
    print()
```

## 2. Which genera are unique to which dataset, and which overlap?

```python
non_data_cols = {"sample_id", "group", "age", "country", "health_status", "Other"}

synthetic_genera = set(synthetic.columns) - non_data_cols
real_genera = set(real.columns) - non_data_cols
toy_species_cols = set(toy.columns) - non_data_cols
# toy dataset is species-level (Genus_species); reduce to genus for a fair comparison
toy_genera = {c.split("_")[0] for c in toy_species_cols}

print("in ALL three (by genus name):", synthetic_genera & real_genera & toy_genera)
print("in example_gut_samples + real_gut_usa_malawi only:", (synthetic_genera & real_genera) - toy_genera)
print("in toy_species_abundance only:", toy_genera - synthetic_genera - real_genera)
```

## 3. One of the two, visualized across all three

`Faecalibacterium` and `Bifidobacterium` both turned up everywhere —
this section picks `Bifidobacterium` since notebook 06 already used it
as the running example. (**YOUR TURN #1** below asks you to check
`Faecalibacterium` too.)

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 3, figsize=(11, 4), sharey=False)

synthetic.boxplot(column="Bifidobacterium", by="group", ax=axes[0])
axes[0].set_title("example_gut_samples\n(synthetic, by mood group)")

real.boxplot(column="Bifidobacterium", by="country", ax=axes[1])
axes[1].set_title("real_gut_usa_malawi\n(real, by country)")

toy.boxplot(column="Bifidobacterium_adolescentis", by="health_status", ax=axes[2])
axes[2].set_title("toy_species_abundance\n(synthetic, by health status)")

fig.suptitle("")
plt.tight_layout()
plt.show()
```
