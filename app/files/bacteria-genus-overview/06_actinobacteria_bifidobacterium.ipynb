## Infant gut, probiotics, and human milk

*Bifidobacterium* dominates the gut of breastfed infants, often making up
the majority of their gut community in the first months of life. It
thrives there because it can digest **human milk oligosaccharides
(HMOs)** — complex sugars in breast milk that the infant itself can't
digest, but that *Bifidobacterium* is specially equipped to ferment. It's
also the single most common genus used in commercial probiotics (species
like *B. longum*, *B. infantis*, *B. animalis* show up on yogurt labels
regularly).

## 1. Bifidobacterium, three datasets, three column names

Notice the resolution difference: two datasets stop at genus
(`Bifidobacterium`), one resolves to species (`Bifidobacterium_adolescentis`)
— the exact distinction notebook 01 introduced.

```python
import pandas as pd

c1_synthetic = pd.read_csv("../gut-microbiome-under-glass/example_gut_samples.csv")
c1_real = pd.read_csv("../gut-microbiome-under-glass/real_gut_usa_malawi.csv")
c2_toy = pd.read_csv("../bioinformatics-of-gut-health/toy_species_abundance.csv")

print("course 1 synthetic, by group:")
print(c1_synthetic.groupby("group")["Bifidobacterium"].mean().round(2))
print()
print("course 1 real, by country:")
print(c1_real.groupby("country")["Bifidobacterium"].mean().round(2))
print()
print("course 2 toy, by health_status (species-level column):")
print(c2_toy.groupby("health_status")["Bifidobacterium_adolescentis"].mean().round(2))
```
