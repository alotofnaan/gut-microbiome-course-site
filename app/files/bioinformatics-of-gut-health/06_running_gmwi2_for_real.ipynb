## The real tool

**GMWI2** (Gut Microbiome Wellness Index 2) — Chang, Gupta *et al.* (2024),
*Nature Communications*, from Mayo Clinic's Sung Lab. Trained on 8,069
pooled stool shotgun metagenomes from 54 studies across 26 countries.
Open-source, installable via `bioconda`.

**Citation:** [doi:10.1038/s41467-024-51651-9](https://doi.org/10.1038/s41467-024-51651-9)
&middot; **Code:** [github.com/danielchang2002/GMWI2](https://github.com/danielchang2002/GMWI2)

## Install and run, end to end

```bash
# one-time setup
mamba install -c bioconda -c conda-forge gmwi2=1.6

# run on one sample's raw paired-end reads
gmwi2 -f forward.fastq -r reverse.fastq -n 8 -o output_prefix
```

Under the hood, this one command runs the entire chain you've now built
by hand across notebooks 01, 02, 03, and 05: QC, human-read removal,
MetaPhlAn3 profiling, presence/absence conversion, and scoring with the
pre-trained Lasso logistic regression model — the same model class you
trained yourself in notebook 05, just fit on 8,069 real samples instead
of 80 synthetic ones.

```
output_prefix_GMWI2.txt          <- the score
output_prefix_GMWI2_taxa.txt     <- which taxa in THIS sample drove the score
output_prefix_metaphlan.txt      <- the raw MetaPhlAn3 species table (notebook 02's output)
```

## The shortcut: scoring an existing MetaPhlAn profile

If MetaPhlAn3 has already been run (notebook 02's output exists), GMWI2
skips straight to scoring using the same pre-trained model — this step
really is just Python + a saved scikit-learn model, conceptually
identical to calling `.predict()` on the model you trained in notebook 05:

```bash
wget https://raw.githubusercontent.com/danielchang2002/GMWI2/main/src/gmwi2_metaphlan_output.py
wget https://raw.githubusercontent.com/danielchang2002/GMWI2/main/src/GMWI2/GMWI2_databases/GMWI2_model.joblib

python3 gmwi2_metaphlan_output.py metagenome_metaphlan_output.txt GMWI2_model.joblib output_prefix
```
