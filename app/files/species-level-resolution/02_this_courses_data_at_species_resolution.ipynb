# This Course's Data, at Species Resolution

Course 3 (Bacteria Genus Overview) looked at several of these same
organisms — but always collapsed to genus. This notebook looks at the
exact same underlying genera, one resolution level down, using columns
that were sitting in the bioinformatics course's dataset the whole time.

## 1. Genus-level vs. species-level, side by side

```python
import pandas as pd

toy = pd.read_csv("../bioinformatics-of-gut-health/toy_species_abundance.csv")
species_cols = [c for c in toy.columns if c not in ("sample_id", "health_status")]
genus_of = {c: c.split("_")[0] for c in species_cols}
print("species column -> genus:")
for sp, g in genus_of.items():
    print(f"  {sp:32s} -> {g}")
```

## 2. How many genera does this dataset actually resolve to species?

Every column here is already species-level — but it's worth checking
whether any TWO columns collapse to the SAME genus (which would mean this
dataset actually distinguishes between species within one genus, not just
across genera).

```python
from collections import Counter
genus_counts = Counter(genus_of.values())
print(genus_counts)
```
