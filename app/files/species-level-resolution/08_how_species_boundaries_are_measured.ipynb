# How Species Boundaries Are Actually Measured

Notebook 01 named ANI. This notebook covers the actual toolkit — what a
lab or bioinformatics pipeline runs to decide "same species" or "different
species" in practice.

## Whole-genome comparison: ANI in practice

Tools like **FastANI** compute average nucleotide identity between two
assembled genomes directly, fragment by fragment. This requires having
full genome assemblies for both organisms being compared — which in turn
requires either isolating and sequencing pure cultures, or assembling
genomes out of shotgun metagenomic data (course 2's pipeline), a much
harder computational problem than the taxonomic *profiling* course 2
actually covered.

## MLST: a lighter-weight compromise

**Multi-Locus Sequence Typing (MLST)** sequences a small, fixed set of
"housekeeping" genes (typically 7) present in every member of a species,
and compares just those. It's much cheaper than whole-genome comparison
and has been used for decades — including for real-world outbreak
tracking (identifying which *E. coli* pathotype, notebook 06, caused a
specific outbreak) — at the cost of resolution: two genomes can share
identical MLST profiles while differing substantially elsewhere in the
genome.

## Marker-gene methods (16S, MetaPhlAn3) are a different tool entirely

Worth being precise about the comparison: 16S sequencing (courses 1, 3)
and MetaPhlAn3-style marker-gene profiling (course 2) were never designed
to do what ANI/MLST do. They're built for **fast, cheap community
profiling** across many samples — trading resolution for throughput. ANI
and MLST are built for **confirming identity of one organism** at high
resolution. Real pipelines often use both: marker genes to survey a whole
community cheaply, then whole-genome methods when a specific organism
needs confirming.
