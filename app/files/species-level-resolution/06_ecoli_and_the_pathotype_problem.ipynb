# *Escherichia coli* and the Pathotype Problem

Every other notebook in this course has been about splitting one
species name into several. This one is the opposite, deliberately: ONE
species, *Escherichia coli*, containing behavior so different that
species-level resolution *itself* stops being enough — which is exactly
why a future course on this site covers strains specifically.

## One species, wildly different outcomes

The genus course's notebook 07 already flagged this. To make it concrete:
*E. coli* **pathotypes** are groups of strains sharing specific
virulence genes:

- **EHEC** (enterohemorrhagic *E. coli*, e.g. the well-known O157:H7) —
  produces Shiga toxin, causes severe illness, occasional large outbreaks
  traced to contaminated food.
- **UPEC** (uropathogenic *E. coli*) — the leading cause of urinary tract
  infections.
- **Commensal *E. coli*** — the vast majority of *E. coli* actually
  living in a healthy gut, causing no disease at all.

All of these are the same species by the ANI convention from notebook
01. Species-level sequencing data — even perfect species-level data —
genuinely cannot tell these apart. That requires strain-level typing
(whole-genome comparison of specific virulence genes, or methods like
MLST from notebook 08).

## 1. What our data can say — and its real limit

```python
import pandas as pd

toy = pd.read_csv("../bioinformatics-of-gut-health/toy_species_abundance.csv")
toy.groupby("health_status")["Escherichia_coli"].describe()[["mean", "std", "min", "max"]].round(2)
```

Whatever this column shows, it cannot distinguish "elevated commensal
*E. coli*, probably harmless" from "the presence of a dangerous
pathotype" — those are indistinguishable at this resolution, by
construction, not by a limitation of this particular dataset.
