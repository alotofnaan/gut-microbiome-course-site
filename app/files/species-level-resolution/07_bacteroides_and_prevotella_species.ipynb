## *Bacteroides* species

The genus course's notebook 02 covered *Bacteroides* the genus as a
diet-driven fiber/protein specialist. Species-level, the genus splits
into organisms with real functional differences:

- ***B. thetaiotaomicron*** — has one of the largest known repertoires of
  genes for breaking down complex plant carbohydrates (glycan-foraging
  genes); a favorite model organism for exactly this reason.
- ***B. fragilis*** — usually a harmless minority species, but the one
  most often implicated when *Bacteroides* causes an actual infection
  (typically after gut contents escape the intestine, e.g. after a
  ruptured appendix). Also produces a well-studied capsular
  polysaccharide (PSA) that interacts with the immune system in ways
  actively researched for both harm and potential benefit.
- ***B. vulgatus*** — typically one of the most abundant *Bacteroides*
  species by sheer numbers in Western-diet guts.

## *Prevotella* species

The genus course's notebook 02 also introduced *Prevotella copri* and its
studied (correlational) association with rheumatoid arthritis in some
cohorts. That research is specifically about *P. copri* — later genomic
work has shown "*P. copri*" itself may represent a complex of several
closely related clades, an echo of exactly the *Faecalibacterium*
situation from notebook 03 of this course.
