## Presence isn't activity

Every gene in this course's dataset is recorded as present or absent.
But a gene being **present in the genome** and a gene being **actively
used** at a given moment are different facts. Bacteria regulate which
genes get transcribed based on their environment — a phenomenon as basic
as turning genes on and off in response to conditions, and as
specific as **phase variation**, where some bacteria randomly and
reversibly switch specific genes on/off across generations, producing
genetically identical daughter cells with different active traits.

## Why this matters concretely

A strain carrying `stx1`/`stx2`/`eae` (notebook 03) is *capable* of being
dangerous — but toxin genes that are present but not being expressed at
meaningful levels, in a specific host, at a specific time, may matter
less in practice than the simple presence/absence table suggests. Real
clinical and research microbiology increasingly pairs genome sequencing
(what CAN this organism do) with **gene expression measurement** — RNA
sequencing, not DNA sequencing — to ask what it's actually doing right
now. That's a different measurement technology than anything covered
across all five courses on this site.
