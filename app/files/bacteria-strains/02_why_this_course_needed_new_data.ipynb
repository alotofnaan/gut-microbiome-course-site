# Why This Course Needed New Data

Every other course on this site reused an existing dataset (course 3 and
4 both read CSVs from earlier courses via relative paths). This one
couldn't — worth explaining exactly why before using the new data.

## The gap, precisely

- `example_gut_samples.csv` and `real_gut_usa_malawi.csv`: genus-level only.
- `toy_species_abundance.csv`: species-level — but it tracks presence and
  *abundance* of named species, not which specific genes any one strain
  of those species carries.

None of the three can show two strains of the *same* species differing
in a specific trait, because none of them were built with that kind of
column in the first place. This isn't a lucky coincidence — it's the
same story as the species course's notebook 07 (*Bacteroides*/*Prevotella*
species): sometimes the honest move is new, clearly-labeled synthetic
data rather than forcing an existing dataset to answer a question it
wasn't built for.

gene columns — five real virulence genes plus one real antibiotic

illustrative**, same as the bioinformatics course's teaching dataset —
real gene names and real biological associations, invented specific
numbers, calibrated so a classifier (notebook 09) lands at a realistic,
imperfect accuracy rather than a suspiciously clean one.
