# Proteobacteria: *Escherichia* & *Klebsiella*

A phylum-level shift in tone. Most genera so far are quietly beneficial;
Proteobacteria is the phylum most associated with things going wrong. A
Proteobacteria "bloom" — a sudden increase in relative abundance — is one
of the more consistent signatures of dysbiosis across many different
conditions.

## *Escherichia* (species: *coli*)

Mostly a harmless, even useful, facultative anaerobe — it can tolerate
the small amount of oxygen near the gut wall that strict anaerobes can't,
carving out its own niche. A small number of specific *E. coli* strains
(like the well-known O157:H7) produce toxins and cause serious illness —
same genus, same species even, wildly different clinical picture depending
on strain. (Exactly the notebook 11 theme, arriving early.)

## *Klebsiella* (species: *pneumoniae*)

Normally a minor, low-abundance gut resident. Clinically significant
because *Klebsiella pneumoniae* is one of the pathogens most associated
with **carbapenem-resistant** infections — carbapenems are often the
"last-resort" antibiotic class, and resistant *Klebsiella* is a real,
ongoing public-health concern, not a hypothetical one.
