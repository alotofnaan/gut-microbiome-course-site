## 1. What is "the gut microbiome," actually?

Your large intestine hosts on the order of **tens of trillions of microbial
cells** — roughly the same order of magnitude as the number of human cells in
your entire body. Collectively they carry **hundreds of times more genes**
than the human genome does.

Most of what we measure is **bacteria**, though the microbiome also includes
archaea, fungi, and viruses. When a paper says "gut microbiome," by default
assume it means "the bacterial community living in stool/colon samples,"
because that's almost always what got sequenced.

## 2. The big picture: phyla

Zoom all the way out and gut bacteria sort into a handful of phyla. Two
dominate in most humans:

- **Firmicutes** — includes *Faecalibacterium*, most *Ruminococcus* and
  *Blautia* — generally fiber-fermenters and short-chain-fatty-acid producers.
- **Bacteroidetes** — includes *Bacteroides* and *Prevotella* — also major
  fiber degraders, but with different preferred substrates and very different
  ecological "personalities" (more on this in Section 4).

Their ratio (the "Firmicutes/Bacteroidetes ratio") got a lot of attention in
early obesity research. Treat that literature skeptically — it's a good
example of a real biological axis being oversimplified into a single number.
Smaller phyla worth knowing: **Actinobacteria** (includes *Bifidobacterium*)
and **Proteobacteria** (includes *Escherichia*) — usually minor by abundance,
but disproportionately important as opportunists when something goes wrong.

## 3. Meet the genera you've already been looking at

Every CSV in this course has columns named after these. Here's what each one
actually does:

| Genus | Phylum | What it does |
|---|---|---|
| *Bacteroides* | Bacteroidetes | Major fiber/carbohydrate degrader; dominant in high-protein, low-fiber ("Western") diets |
| *Prevotella* | Bacteroidetes | Also a fiber degrader, but thrives on **plant-heavy, high-fiber** diets; classically the mirror image of *Bacteroides* across populations |
| *Faecalibacterium* | Firmicutes | Produces **butyrate**, a short-chain fatty acid that feeds colon cells and has anti-inflammatory effects; low levels are associated with several gut disorders |
| *Bifidobacterium* | Actinobacteria | Dominant in breastfed infants; commonly used as a probiotic strain; generally considered beneficial |
| *Escherichia* | Proteobacteria | Usually a small, harmless fraction — but blooms are a classic signature of inflammation or dysbiosis |

## 4. The gut-brain axis

The `low_mood` vs `control` grouping in this course's teaching dataset isn't
an arbitrary label — it's a nod to a real and active research area: the
**gut-brain axis**, the bidirectional communication network between gut
microbes and the central nervous system. Three mechanisms come up constantly:

1. **The vagus nerve** — a direct physical wire between gut and brain that
   microbial signals can influence.
2. **Neurotransmitter precursors** — some gut bacteria produce or modulate
   levels of compounds related to serotonin and GABA synthesis.
3. **Inflammation** — a disrupted, low-diversity microbiome can increase gut
   permeability ("leaky gut") and systemic inflammation, which is separately
   linked to mood disorders.

Important caveat for a skeptical scientist-in-training: most human evidence
here is **correlational**, sample sizes are often small, and causality (does
the microbiome affect mood, does mood affect the microbiome, or both?) is
still an open question. This is exactly the kind of claim that notebooks 03
and 04 will teach you to interrogate rather than accept at face value.

## 5. Diet, geography, and Prevotella vs Bacteroides

The single most reproducible finding in human gut microbiome research is a
diet effect: populations eating **high-fiber, plant-heavy** diets run high
*Prevotella* and low *Bacteroides*; populations eating **low-fiber, high-fat
"Western"** diets show the reverse. This isn't a subtle statistical
squint — it's one of the largest, most robust effect sizes in the field.

You'll reproduce this exact finding on real published data in notebook 05

## 6. Dysbiosis and diversity

**Dysbiosis** is the general term for "the microbial community is out of its
healthy range" — usually meaning lower diversity, loss of key beneficial
species (like *Faecalibacterium*), and/or blooms of normally-minor
opportunists (like *Escherichia*). It's associated with (not proven to
cause) a long list of conditions: inflammatory bowel disease, obesity,
antibiotic aftermath, and more.

"Diversity" so far has just been a word. Notebook 06 turns it into an actual
number you compute yourself — alpha diversity (how rich is one sample?) and
beta diversity (how different are two samples/groups from each other?).

## 7. What these CSVs actually measure

One more thing worth being precise about before you touch more data: none of
these numbers are literal bacteria counts. They come from **16S rRNA gene
sequencing** — a method that sequences one specific, highly conserved gene
present in all bacteria, then groups similar sequences into either:

- **OTUs** (Operational Taxonomic Units) — sequences clustered by similarity
  (e.g. 97% identical), an older approach, or
- **ASVs** (Amplicon Sequence Variants) — exact-sequence resolution, the
  modern standard (notebook 07 explains why ASVs replaced OTUs).

Each OTU/ASV is then counted and turned into a **relative abundance** — the
percentage of *sequencing reads* assigned to that group, not an absolute cell
count. This matters: if one bacterium doubles in absolute number but
everything else doubles too, its relative abundance stays flat. Relative
abundance data is also **compositional** — all values in a row sum to ~100%,
so genera aren't statistically independent of each other. Keep that in the
back of your mind; it's part of why microbiome statistics has its own
specialized methods beyond a plain t-test.
