# Is "Species" a Natural Category for Bacteria?

Every notebook so far has treated species as a real thing to be
discovered, just hard to measure precisely. This notebook asks a harder
question: is it fully a real, discoverable category at all, or partly a
convenient filing system humans imposed on something messier?

## Why bacteria break the usual species logic

The biological species concept (interbreeding populations, used for
animals and plants) assumes sexual reproduction creates a natural
boundary. Bacteria reproduce asexually by dividing, and separately swap
genes across what look like species and even genus lines via
**horizontal gene transfer** (HGT) — a gene from one species can end up
functioning inside an entirely different one. That means the tidy
"reproductively isolated population" logic underlying animal species
doesn't cleanly apply.

## The field's actual response: clustering, not discovery

This is exactly why notebook 01's ANI threshold (~95-96%) is described as
a **convention** rather than a **discovery**. Genomes really do cluster —
data from real bacterial isolates isn't smoothly, uniformly spread across
all possible similarity values, it does show denser clumps and sparser
gaps. But WHERE exactly to draw the species line within a real, messy
distribution is a judgment call the field standardized for continuity
with pre-genomic taxonomy (notebook 01), not a boundary nature drew for
us.

## What this means for everything in this course

- **The *Faecalibacterium* split (notebook 03)** exists precisely because
  where the ANI line falls is a judgment call, applied to real new genomic
  data.
- **The *Prevotella copri* complex (notebook 07)** is the same situation.
- **The *E. coli* pathotype problem (notebook 06)** shows the flip side —
  organisms that stay on the SAME side of the species line despite huge
  functional differences.

None of this means species-level analysis is useless — it means treating
a species name as a hard, settled fact (rather than the field's current
best clustering convention) is itself a small act of overconfidence,
worth noticing.
