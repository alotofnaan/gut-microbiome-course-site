# *E. coli* Pathotypes, For Real

The genus course and species course both mentioned "EHEC" and "O157:H7"
in passing. This notebook explains what those genes in the dataset
actually do.

## The genes

- **`stx1` / `stx2`** — Shiga toxin genes. These are what make
  enterohemorrhagic *E. coli* (**EHEC**, the O157:H7 group is the most
  famous member) dangerous: the toxin damages blood vessel linings and can
  cause severe illness, including kidney failure in serious cases. Public
  health labs investigating a real foodborne outbreak specifically test
  for these genes by PCR — this is a real, standard diagnostic method,
  not a hypothetical.
- **`eae`** — intimin, a gene that helps the bacterium attach tightly to
  the gut lining. Common in EHEC alongside the toxin genes.
- **`papC`** — P fimbriae, adhesion structures that help a strain grip the
  urinary tract lining. A defining marker of uropathogenic *E. coli*
  (**UPEC**), the single most common cause of urinary tract infections.
- **`hlyA`** — hemolysin, a toxin that damages host cells; also common in
  UPEC strains.

## 1. See the two pathotype signatures in the real (synthetic) data

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("ecoli_strain_profiles.csv")
genes = ["stx1", "stx2", "eae", "papC", "hlyA"]

means = df.groupby("pathotype")[genes].mean()
fig, ax = plt.subplots(figsize=(7, 4))
means.T.plot(kind="bar", ax=ax)
ax.set_ylabel("Fraction of strains carrying the gene")
ax.set_title("Virulence gene signatures by pathotype")
plt.xticks(rotation=0)
fig.tight_layout(); plt.show()
```
