## *Alistipes*

Bile-tolerant, generally more abundant on higher-fat diets. Research on
*Alistipes* and mood/anxiety exists in animal models and some human
cohorts, with directionally mixed results across studies — a good
reminder that "linked to mood" (notebook 01 of the biology course) doesn't
mean "linked the same way in every study."

## *Parabacteroides*

Closely related to *Bacteroides* (the genus name literally means "beside
*Bacteroides*"). One species, *Parabacteroides distasonis*, has drawn
research interest for potential metabolic benefits (studied in mouse
models for effects on weight and blood sugar) — promising, still early,
still mostly animal-model evidence.

## 1. Two genera, two different datasets

Worth noticing before writing any code: `Parabacteroides` is a column in
the biology course's real USA/Malawi data, but `Alistipes` isn't — it
only shows up (as `Alistipes_putredinis`, species-level) in the
bioinformatics course's teaching dataset. Real datasets rarely happen to
contain exactly the genera you want to compare side by side; checking
column names before writing analysis code is a real habit, not a
formality.

```python
import pandas as pd

real = pd.read_csv("../gut-microbiome-under-glass/real_gut_usa_malawi.csv")
real.groupby("country")["Parabacteroides"].mean().round(2)
```

```python
toy = pd.read_csv("../bioinformatics-of-gut-health/toy_species_abundance.csv")
toy.groupby("health_status")["Alistipes_putredinis"].mean().round(2)
```
