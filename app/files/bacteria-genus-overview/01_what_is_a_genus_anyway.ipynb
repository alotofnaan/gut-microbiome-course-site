# What Is a Genus, Anyway?

Before profiling individual genera, it's worth being precise about what
the word means and where these names actually come from — including one
small mystery already sitting in a dataset you may have used before.

## 1. The rank hierarchy

Bacterial taxonomy is organized into nested ranks, most specific last:

```
Domain → Phylum → Class → Order → Family → Genus → Species
```

*Faecalibacterium prausnitzii* — the butyrate producer from the other
courses — sits at:

```
Bacteria → Firmicutes → Clostridia → Eubacteriales → Oscillospiraceae → Faecalibacterium → prausnitzii
```

**Genus** groups closely related species; **species** (in the two-word
"binomial" name) is the more specific identifier. When a dataset column is
just `Faecalibacterium` (genus-level, no species), it means the underlying
method — usually 16S sequencing — couldn't resolve further than genus.
That's a real limitation, not a rounding choice, and it's exactly why
notebook 11 closes this course on it.

## 2. Where genus names actually come from

Nobody manually names each sequence. Tools like the ones in the
bioinformatics course (QIIME2/DADA2, MetaPhlAn3) match a sequence against
a **reference database** — SILVA, Greengenes, or similar — that has
already assigned names to reference sequences based on prior sequencing,
culturing, and phylogenetic analysis. A "genus" is, practically speaking,
whatever level of agreement a reference database can confidently assign to
a new sequence.

## 3. The mystery bracket: `[Ruminococcus]`

If you've looked closely at `real_gut_usa_malawi.csv` from the other
course, one column is literally named `[Ruminococcus]` — square brackets
and all. Let's look.

````python
import pandas as pd
df = pd.read_csv("../gut-microbiome-under-glass/real_gut_usa_malawi.csv")
print([c for c in df.columns if "Ruminococcus" in c])
````

That's not a typo. In the Greengenes reference database (one of the
standard 16S references), square brackets around a genus name mean:
*taxonomically provisional* — sequences that resemble the genus closely
enough to group with it, but that the database maintainers flag as likely
to be reclassified once better data exists. `Ruminococcus` (no brackets)
and `[Ruminococcus]` (bracketed) are being tracked as **two distinct
columns** in this exact dataset precisely because they're not confidently
the same thing. You'll see genuinely different numbers for each in
notebook 05.
