# Plasmids and Mobile Genetic Elements

Every gene discussed so far in this course has been described as if it
just sits on the bacterium's main chromosome. Many of the most
consequential strain-defining genes don't.

## What a plasmid is

A **plasmid** is a small, separate loop of DNA that replicates
independently of a bacterium's main chromosome. Bacteria can carry zero,
one, or several plasmids, and — critically — **plasmids can transfer
between cells**, including between different species, through a process
called conjugation. A gene sitting on a plasmid isn't tied to one
lineage's family tree the way a chromosomal gene is; it can jump.

## Why this matters for everything in this course

This is the mechanical explanation behind the species course's notebook
09 (horizontal gene transfer complicating the bacterial species concept).
It's also directly relevant to this course's own data: **antibiotic
resistance genes are disproportionately likely to live on plasmids**
rather than the chromosome — which is exactly why resistance can spread
so fast between unrelated strains and even unrelated species, and exactly
why this course's `blaKPC` column was deliberately built to vary somewhat
independently of pathotype (notebook 02) — resistance and virulence
often really do travel on separate genetic elements, with separate
transmission dynamics.

## 1. Check the independence again, explicitly

```python
import pandas as pd

df = pd.read_csv("ecoli_strain_profiles.csv")
overall_rate = df["blaKPC"].mean()
by_pathotype = df.groupby("pathotype")["blaKPC"].mean()
print(f"overall blaKPC rate: {overall_rate:.2f}")
print(by_pathotype.round(2))
```
