# *Bifidobacterium* Species

The genus course's notebook 06 covered *Bifidobacterium* the genus:
dominant in breastfed infants, mostly absent by comparison in adults.
That framing already hid a species-level story — "*Bifidobacterium*"
in an infant gut and "*Bifidobacterium*" in an adult gut are rarely even
the same species.

## The species that actually show up

- ***B. longum*** (including the subspecies *infantis*) — the classic
  infant-gut dominant species, specialized for human milk oligosaccharide
  (HMO) digestion.
- ***B. breve*** — another common infant-associated species.
- ***B. adolescentis*** — despite the name (Latin for "adolescent"), this
  is the species most commonly detected in **adult** guts, not
  teenagers specifically — an old naming choice, not a literal age claim.
  This is the exact species in this course's dataset.

## 1. Our data's species in context
