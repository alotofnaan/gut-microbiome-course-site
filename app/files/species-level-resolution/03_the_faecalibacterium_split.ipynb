# The *Faecalibacterium* Split

For years, "*Faecalibacterium prausnitzii*" was treated as a single
species and the genus's only well-known member — the butyrate producer
introduced in the genus course's notebook 04. More recent whole-genome
comparisons have complicated that picture.

## The real story

Whole-genome sequencing of many *Faecalibacterium* isolates has shown
that what used to be lumped together as "*F. prausnitzii*" actually spans
multiple genome clusters below the ~95% ANI species threshold (notebook
01) from each other — genomically distinct enough, by the field's own
convention, to be considered separate species. Taxonomists have proposed
splitting the old single species into several new ones. This is an
active area of ongoing reclassification, not a single settled paper —
exactly the kind of "the reference database will look different in five
years" situation notebook 01 of the genus course warned about.

## Why this matters beyond bookkeeping

If "*F. prausnitzii*" abundance actually reflects several genomically
distinct organisms lumped under one legacy name, then a study reporting
"*F. prausnitzii* is depleted in Crohn's disease" (genus course, notebook
04) might really mean "one or two of these several species are depleted,
and others aren't" — a more precise, more useful, and more complicated
finding hiding inside a simpler-sounding one.
