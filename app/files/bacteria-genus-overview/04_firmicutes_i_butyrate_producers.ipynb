# Firmicutes I: The Butyrate Producers

Three genera, one shared job: producing **butyrate**, a short-chain fatty
acid (SCFA) made by fermenting fiber the human gut can't digest directly.
Butyrate is the preferred fuel source for colonocytes (the cells lining
your colon) and has documented anti-inflammatory effects. Low levels of
butyrate producers show up repeatedly in inflammatory bowel disease
research.

## *Faecalibacterium* (species: *prausnitzii*)

Usually one of the single most abundant commensal species in a healthy
human gut. Reduced *Faecalibacterium prausnitzii* is one of the most
replicated findings in Crohn's disease research — replicated enough that
it's become a candidate for live-biotherapeutic (next-generation
probiotic) development.

## *Roseburia* and *Eubacterium* (species: *rectale*)

*Roseburia* and *Eubacterium rectale* are so phylogenetically close that
older taxonomies sometimes grouped them together — a real example of
notebook 01's point about genus boundaries shifting as reference data
improves. Both are fiber-responsive: their abundance tends to rise when
dietary fiber intake rises, over a timescale of days.

## 2. Do *Roseburia* and *Eubacterium* move together?

If two genera fill a similar ecological role, you might expect samples
high in one to often be high in the other too. Let's check, using the
bioinformatics course's teaching dataset (the only one with both).

```python
toy = pd.read_csv("../bioinformatics-of-gut-health/toy_species_abundance.csv")
correlation = toy["Roseburia_intestinalis"].corr(toy["Eubacterium_rectale"])
print(f"correlation: {correlation:.2f}")

import matplotlib.pyplot as plt
fig, ax = plt.subplots(figsize=(5, 4))
ax.scatter(toy["Roseburia_intestinalis"], toy["Eubacterium_rectale"], alpha=0.6)
ax.set_xlabel("Roseburia_intestinalis"); ax.set_ylabel("Eubacterium_rectale")
fig.tight_layout(); plt.show()
```
